# dict 和 json 的区别？​
dict 是 Python 里的数据结构，JSON 是一种跨语言传输和存储数据的文本格式；JSON 里的 object 读到 Python 后通常会变成 dict。​
> dict 用于处理内部数据。json 用于跨应用传输数据。

| 对比点 | Python dict | JSON |
| --- | --- | --- |
| 本质 | Python 内置数据结构 | 文本数据格式 |
| 所属 | Python 语言 | 跨语言通用 |
| 是否能直接操作 | 可以直接操作 | 需要解析后才能操作 |
| 常见用途 | 程序内部保存和处理数据 | 网络传输、配置文件、接口数据 |
| Java 类比 | HashMap / LinkedHashMap | JSON 字符串 |

```python
# json字符串
json_text = '{"name": "Alice", "age": 18, "isStudent": true}'
print(type(json_text))
print(json_text)
```

```python
# dict对象
user = {
    "name": "Alice",
    "age": 18
}
print(type(user))
print(user)
```

> PS: json 一段符合 JSON 规则的文本（字符串），单引号包裹，不是对象。在打印 json 的时候，属性是双引号。dict是对象，拥有自己的方法，属性。打印 dict 时，属性是单引号。True 是 Dict，true 是 json。

| Python 类型 | JSON 类型 |
| --- | --- |
| dict | object |
| list / tuple | array |
| str | string |
| int / float | number |
| True | true |
| False | false |
| None | null |

# JSON 和 dict 如何互相转换？
Python 提供了标准库 json 来处理 JSON。
1. JSON 字符串转 dict：json.loads()

```python
import json

json_text = '{"name": "Alice", "age": 18, "isStudent": true}'

# json str 转 dict
user = json.loads(json_text)

print(json_text)
print(f"type(json_text)={type(json_text)}")
print(user)
print(f"type(user)={type(user)}")
print(f'user["name"]={user["name"]}')
```

2. dict 转 JSON 字符串：json.dumps()

```python
import json

user = {
    "name": "Alice",
    "age": 18,
    "isStudent": True
}

json_text = json.dumps(user)

print(json_text)
print(type(json_text))
```

> 有元祖的dict无法转换成json。TypeError: keys must be str, int, float, bool or None, not tuple

```python
import json

data = {    1: "one",    2: "two",    (3, 4): "point"}
print(data)
print(data[(3, 4)])
print(json.dumps(data))
```
