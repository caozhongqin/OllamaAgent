Python 3.11 的 list 类似 Java 1.8 的 ArrayList，dict 类似 Java 1.8 的 HashMap；它们都是常用容器，但 Python 更动态、更灵活，Java 更强调类型约束和编译期检查。​

list 是一种有序、可变、可重复的序列容器。

| 操作 | Python 3.11 list | Java 1.8 ArrayList |
| --- | --- | --- |
| 创建 | nums = [1, 2, 3] | ArrayList<Integer> nums = new ArrayList<>(); |
| 添加元素 | nums.append(4) | nums.add(4) |
| 按下标取值 | nums[0] | nums.get(0) |
| 修改元素 | nums[0] = 100 | nums.set(0, 100) |
| 删除元素 | nums.remove(100) | nums.remove(Integer.valueOf(100)) |
| 长度 | len(nums) | nums.size() |

```python
nums = [10, 20, 30, 30]
print(nums)
print(nums[0])

# 获取元素个数
print(f"len(nums)->{len(nums)}")

# 追加值
nums.append(40)
print(f"nums.append(40)->{nums}")

# 修改值
nums[1]=100
print(f"nums[1]=100->{nums}")

# for-each遍历
print("for-each遍历")
for num in nums:
    print(f"    {num}")

# 删除值（删除第一次匹配到的值）
nums.remove(30)
print(f"nums.remove(30)->{nums}")
# 删除值（删除第一次匹配到的值）
nums.remove(30)
print(f"nums.remove(30)->{nums}")
```

Python 的 list 可以放不同类型的数据

```python
datas = [1, "Hello World", True, 3.14]
print(datas)
for data in datas:
    print(f"    {data}")
```

!!! Python 3.11 的 dict 会保持插入顺序。而 Java 1.8 的 HashMap 不保证插入顺序。

| 操作 | Python 3.11 dict | Java 1.8 HashMap |
| --- | --- | --- |
| 创建 | user = {"name": "Alice"} | HashMap<String, String> user = new HashMap<>(); |
| 添加键值对 | user["age"] = 18 | user.put("age", 18) |
| 取值 | user["name"] | user.get("name") |
| 安全取值 | user.get("name") | user.get("name") |
| 删除 | del user["age"] | user.remove("age") |
| 判断 key 是否存在 | "name" in user | user.containsKey("name") |
| 遍历 key | for k in user: | for (String k : user.keySet()) |
| 遍历 value | for v in user.values(): | for (Object v : user.values()) |
| 遍历键值对 | for k, v in user.items(): | for (Map.Entry<K,V> e : map.entrySet()) |

```python
user = {
    "name": "Alice",
    "age": 18,
    "city": "Beijing"
}

print(user["name"])  # Alice
print(user["age"])   # 18
```
